```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import SparsePauliOp
from qiskit.primitives import Estimator
from qiskit_algorithms import VQE
from qiskit_algorithms.optimizers import COBYLA
from qiskit.circuit.library import TwoLocal
```

```python
mu2 = 1.0

D = np.array([
    [ 5.26786, -3.39048,  0.87619],
    [-3.39048,  5.26786, -3.39048],
    [ 0.87619, -3.39048,  5.26786]
])

Omega = mu2*np.eye(3) + D
```

```python
## Embed Ω into 4×4
import numpy as np

H_full = np.zeros((4,4))

# physical indices
phys = [0,1,2]

for i in range(3):
    for j in range(3):
        H_full[phys[i], phys[j]] = Omega[i,j]
        
```

```python
## Convert to Qiskit operator

from qiskit.quantum_info import Operator, SparsePauliOp

op = Operator(H_full)
H = SparsePauliOp.from_operator(op)
```

```python
### Quantum circuit

from qiskit import QuantumCircuit
from qiskit.circuit import Parameter

theta = Parameter('θ')
phi   = Parameter('φ')

qc = QuantumCircuit(2)

# Mix |00> <-> |01>
qc.ry(2*theta, 1)

# Controlled rotation only when q1 = 0
qc.x(1)
qc.cry(2*phi, 1, 0)
qc.x(1)

qc.draw('mpl')
```

```python
## Run VQE on Quantum Simulator

from qiskit_algorithms import VQE
from qiskit.primitives import StatevectorEstimator
from qiskit_algorithms.optimizers import COBYLA
from qiskit_aer.primitives import EstimatorV2
from qiskit_algorithms.optimizers import SPSA
optimizer = SPSA(maxiter=300)
result_exact = VQE(StatevectorEstimator(), qc, COBYLA()).compute_minimum_eigenvalue(H)

estimator = EstimatorV2(
    options={"run_options": {"shots": 100000}}
)
vqe = VQE(
    estimator,            # shot-based
    qc,
    optimizer,
    initial_point=result_exact.optimal_point
)

#vqe = VQE(estimator, qc, optimizer)
result = vqe.compute_minimum_eigenvalue(H)

print(result.eigenvalue)
```

```python
bound = qc.assign_parameters(result_exact.optimal_point)

job = estimator.run([(bound, H)])
energy = job.result()[0].data.evs

print("Shot energy at exact params:", energy)
```

```python
import numpy as np

best_energy = +1e9
best_result = None

for k in range(10):  # number of restarts

    init = 2*np.pi*np.random.rand(qc.num_parameters)

    vqe = VQE(
        estimator,
        qc,
        optimizer,
        initial_point=init
    )

    result = vqe.compute_minimum_eigenvalue(H)

    if result.eigenvalue < best_energy:
        best_energy = result.eigenvalue
        best_result = result

print("Best energy:", best_energy)
```

```python
## Compare with Exact
print(np.linalg.eigvalsh(Omega))
print("Exact energy:", np.min(np.linalg.eigvalsh(Omega)))
```

```python
#Definitive Test (no VQE needed)

from qiskit_aer import AerSimulator
from qiskit import transpile

sim = AerSimulator()

def pauli_expectation(qc, pauli_label, shots=100000):
    circ = qc.copy()

    # Rotate to measurement basis
    for i, p in enumerate(pauli_label[::-1]):  # Qiskit bit order
        if p == 'X':
            circ.h(i)
        elif p == 'Y':
            circ.sdg(i)
            circ.h(i)

    circ.measure_all()

    tqc = transpile(circ, sim)
    counts = sim.run(tqc, shots=shots).result().get_counts()

    exp = 0.0
    for bitstring, n in counts.items():

        sign = 1
        for bit, p in zip(bitstring[::-1], pauli_label[::-1]):
            if p != 'I' and bit == '1':
                sign *= -1

        exp += sign * n

    return exp / shots
```

```python
#Compute energy of VQE state

params = result.optimal_point
bound = qc.assign_parameters(params)

E = 0.0
for P, c in zip(H.paulis, H.coeffs):
    exp = pauli_expectation(bound, P.to_label())
    E += c.real * exp

print("Shot energy:", E)
```
