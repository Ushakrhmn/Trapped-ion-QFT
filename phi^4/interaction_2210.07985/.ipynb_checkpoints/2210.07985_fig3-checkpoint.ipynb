```python
import numpy as np
import matplotlib.pyplot as plt
### d.o.f. of one site Boson=64
nq = 6
Nphi = 2**nq

alpha = np.arange(Nphi)
```

```python
### Define phi grid (eq. 11)
def get_phi_grid(mu):
    dphi = np.sqrt(2*np.pi/(Nphi * mu))
    phi = dphi * (alpha - (Nphi - 1)/2)
    return phi, dphi
```

```python
####Build operators Phi and Pi

def build_operators(phi, mu):
    Phi = np.diag(phi)

    F = np.fft.fft(np.eye(Nphi)) / np.sqrt(Nphi)
    Pi = mu * F @ Phi @ F.conj().T

    return Phi, Pi
```

```python
### Build local Hamiltonian (eq. 42)

def build_H(Phi, Pi, mI2, lambdaI, fI):
    H_loc = 0.5 * Pi @ Pi \
        + 0.5 * mI2 * Phi @ Phi \
        + (lambdaI / 24.0) * Phi @ Phi @ Phi @ Phi \
        + fI * Phi
    return H_loc
```

```python
## Solve for ground state

def ground_state(H):
    eigvals, eigvecs = np.linalg.eigh(H)
    return eigvecs[:, 0]
```

```python
## Harmonic oscillator (Black dots)
mI2 = 1.0
lambdaI = 0.0
fI = 0.0
mu=np.sqrt(abs(mI2))
phi, _ = get_phi_grid(mu)
Phi, Pi = build_operators(phi, mu)
H_loc_HO = build_H(Phi, Pi, mI2, lambdaI, fI)

psi_ho = ground_state(H_loc_HO)
```

```python
### Strong interactions (red squares)
mI2 = 1.0
lambdaI = 100.0 * np.sqrt(mI2)**3
fI = 0.0
mu=np.sqrt(abs(mI2))
phi, _ = get_phi_grid(mu)
Phi, Pi = build_operators(phi, mu)
H_loc_strong = build_H(Phi, Pi, mI2, lambdaI, fI)

psi_strong = ground_state(H_loc_strong)
```

```python
#### smaller dphi (green diamonds)
mI2 = 1.0
lambdaI = 100.0 * np.sqrt(mI2)**3
fI = 0.0
mu=np.sqrt(abs(mI2))
mu_eff = 4.0 * mu   # shrinks dphi

phi, _ = get_phi_grid(mu_eff)
Phi, Pi = build_operators(phi, mu_eff)
H_loc_strong_smaller_dphi = build_H(Phi, Pi, mI2, lambdaI, fI)

psi_strong_dense = ground_state(H_loc_strong_smaller_dphi)
```

```python
#### Negative mass (Blue triangle)

mI2 = -1.0
lambdaI = 0.1
fI = 1e-5
mu=np.sqrt(abs(mI2))
mu_eff=mu/4
phi, _ = get_phi_grid(abs(mu_eff))
Phi, Pi = build_operators(phi, abs(mu_eff))
H_loc_neg_mass = build_H(Phi, Pi, mI2, lambdaI, fI)

psi_double = ground_state(H_loc_neg_mass)
```

```python
### negative mass+strong field (magenta squares)
mI2 = -1.0
lambdaI = 0.1
fI = 4
mu=np.sqrt(abs(mI2))
mu_eff=mu/4
phi, _ = get_phi_grid(abs(mu_eff))
Phi, Pi = build_operators(phi, abs(mu_eff))
H_loc_neg_mass_strong = build_H(Phi, Pi, mI2, lambdaI, fI)

psi_double_shifted = ground_state(H_loc_neg_mass_strong)
```

```python
plt.figure(figsize=(6,4))

plt.plot(alpha, np.abs(psi_ho), '-', color='black', label='HO (λ=0)')
plt.plot(alpha, np.abs(psi_strong), '-', color='red', label='Strong λ')
plt.plot(alpha, np.abs(psi_strong_dense), '-', color='green', label='Strong λ (Δ/2)')
plt.plot(alpha, np.abs(psi_double), '-', color='blue', label='Double well')
plt.plot(alpha, np.abs(psi_double_shifted), '-', color='magenta', label='Shifted')

plt.legend()
plt.xlabel(r'$\alpha$')
plt.ylabel(r'$|\psi(\phi_\alpha)|$')
plt.title('Reproducing Fig. 2')
plt.show()
```
